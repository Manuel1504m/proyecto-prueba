```python
# Motor de reglas con enfoque empático (Industria 5.0)

class MotorEmpatia:
    def _init_(self):
        self.reglas = {
            "estres": "Respira profundo 🙏. Tómate una pausa de 5-10 min.",
            "triste": "No estás solo 💙. Considera hablar con alguien de confianza.",
            "cansado": "Tu cuerpo necesita descanso ⚡. Dormir bien es clave.",
            "ansioso": "Prueba la técnica de respiración 4-7-8 🧘.",
            "solo": "Conectar con otros ayuda 🤝. ¿Hay alguien con quien puedas hablar?"
        }

    def detectar_emocion(self, texto):
        texto = texto.lower()
        for palabra in self.reglas:
            if palabra in texto:
                return palabra
        return "neutral"

    def responder(self, emocion):
        return self.reglas.get(
            emocion,
            "Estoy aquí para escucharte 😊. Cuéntame más."
        )


class AsistenteEmpatico:
    def _init_(self):
        self.motor = MotorEmpatia()

    def procesar(self, mensaje):
        emocion = self.motor.detectar_emocion(mensaje)
        respuesta = self.motor.responder(emocion)
        return f"🧠 Emoción detectada: {emocion}\n💬 {respuesta}"
```

```python
import gradio as gr

asistente = AsistenteEmpatico()

def interfaz_usuario(mensaje):
    return asistente.procesar(mensaje)

# Diseño accesible (Industria 5.0)
demo = gr.Interface(
    fn=interfaz_usuario,
    inputs=gr.Textbox(
        placeholder="Escribe cómo te sientes...",
        label="💭 Cuéntame cómo te sientes",
        lines=3
    ),
    outputs=gr.Textbox(label="🤖 Respuesta del asistente"),

    title="🧠 AI Empathy Assistant",

    description="""
    Este asistente inteligente está diseñado para ayudarte con recomendaciones empáticas.

    🌍 Enfoque:
    ✔️ Human-centered (Industria 5.0)
    ✔️ Inclusión digital
    ✔️ Apoyo emocional básico
    """,

    theme="soft",
    allow_flagging="never"
)

demo.launch()
```
